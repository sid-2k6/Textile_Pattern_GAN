# Baseline 2 — Conditional WGAN-GP on Batik_Lasem (canonical 50% subset)

**Project:** Deep GANs for Aesthetic-Driven Apparel Pattern Synthesis.

Conditional **WGAN-GP** generating **128×128 RGB** Batik_Lasem motifs, conditioned
on `motif_name` (5 classes).

* Same canonical 50% subset, same group-aware `origin_images` split, same held-out
  FID/KID protocol as the DCGAN baseline (only the model/algorithm differs).
* **Wasserstein loss + gradient penalty** (λ·(‖∇C(x̂)‖₂ − 1)²). The critic uses
  **InstanceNorm (no BatchNorm)** — BatchNorm invalidates the per-sample GP.
* `N_CRITIC` critic updates per generator update (configurable).
* Trained in **fp32** (mixed precision is unstable with the GP double-backward —
  documented, not silently changed).

> **Metrics note:** a WGAN critic outputs an unbounded score, not a probability,
> so *discriminator accuracy is not defined* here. We therefore report the
> **critic loss, gradient penalty and Wasserstein estimate** instead, and leave
> `real_accuracy`/`fake_accuracy` as N/A (never fabricated). Held-out FID/KID use
> the identical protocol as the other baselines.

## 13. Model definition

```python

# =====================================================================
# MODEL DEFINITION (conditional WGAN-GP generator + critic, 128x128)
# =====================================================================
G, Dnet = models.build_models("cwgan_gp", CONFIG, DEVICE)
print("Generator params:", f"{models.count_parameters(G):,}")
print("Critic params   :", f"{models.count_parameters(Dnet):,}")
CONFIG["generator_parameters"] = models.count_parameters(G)
CONFIG["discriminator_parameters"] = models.count_parameters(Dnet)
MOTIF_ORDER = D.MOTIF_ORDER
```

## 14. Optimizers

```python

# =====================================================================
# OPTIMIZERS (Adam, betas=(0.0,0.9) as usual for WGAN-GP). No GradScaler (fp32).
# =====================================================================
opt_g = torch.optim.Adam(G.parameters(), lr=CONFIG["learning_rate"],
                         betas=(CONFIG["beta1"], CONFIG["beta2"]))
opt_d = torch.optim.Adam(Dnet.parameters(), lr=CONFIG["learning_rate"],
                         betas=(CONFIG["beta1"], CONFIG["beta2"]))
```

## 15. Checkpoint / resume

```python

# =====================================================================
# CHECKPOINT MANAGER + HISTORY + RESUME
# =====================================================================
HISTORY_COLUMNS = ["epoch", "generator_loss", "critic_loss", "gradient_penalty",
                   "wasserstein_estimate", "real_accuracy", "fake_accuracy",
                   "fid", "kid_mean", "kid_std", "diversity",
                   "lr_g", "lr_d", "epoch_time", "gpu_memory_mb"]
ckpt = CheckpointManager(paths.checkpoint_dir,
                         save_interval=CONFIG["checkpoint_interval"], best_mode="min")
history = HistoryLogger(paths.history_csv, HISTORY_COLUMNS)

start_epoch = 0
resume_path = ckpt.resolve_resume(CONFIG["resume"], CONFIG["resume_checkpoint"])
if resume_path:
    state = ckpt.load(resume_path, map_location=DEVICE)
    G.load_state_dict(state["G"]); Dnet.load_state_dict(state["D"])
    opt_g.load_state_dict(state["opt_g"]); opt_d.load_state_dict(state["opt_d"])
    restore_rng_states(state.get("rng"))
    ckpt.best_metric = state.get("best_metric", float("inf"))
    start_epoch = int(state.get("epoch", 0))
    print(f"Checkpoint found. Resuming from epoch {start_epoch}. best FID = {ckpt.best_metric}")
else:
    print("Starting training from epoch 0.")
```

## 16. Training loop

```python

# =====================================================================
# TRAINING LOOP — Conditional WGAN-GP
#   critic:   L = E[C(fake)] - E[C(real)] + lambda_gp * GP
#   generator L = -E[C(fake)]  (updated every N_CRITIC critic steps)
# =====================================================================
def sample_generator(n):
    z = torch.randn(n, CONFIG["latent_dim"], device=DEVICE)
    labels = torch.randint(0, CONFIG["num_classes"], (n,), device=DEVICE)
    return G(z, labels)

for epoch in range(start_epoch + 1, CONFIG["epochs"] + 1):
    G.train(); Dnet.train()
    env.reset_peak_memory()
    t0 = time.time()
    c_losses, g_losses, gps, wass = [], [], [], []
    step = 0

    for real, labels in train_loader:
        real = real.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        bs = real.size(0)

        # ---------- Critic ----------
        opt_d.zero_grad(set_to_none=True)
        z = torch.randn(bs, CONFIG["latent_dim"], device=DEVICE)
        fake = G(z, labels).detach()
        d_real = Dnet(real, labels).mean()
        d_fake = Dnet(fake, labels).mean()
        gp = T.gradient_penalty(Dnet, real, fake, DEVICE, labels=labels)
        loss_d = d_fake - d_real + CONFIG["lambda_gp"] * gp
        loss_d.backward(); opt_d.step()
        c_losses.append(loss_d.item()); gps.append(gp.item())
        wass.append((d_real - d_fake).item())
        step += 1

        # ---------- Generator (every N_CRITIC steps) ----------
        if step % CONFIG["n_critic"] == 0:
            opt_g.zero_grad(set_to_none=True)
            z = torch.randn(bs, CONFIG["latent_dim"], device=DEVICE)
            gen = G(z, labels)
            loss_g = -Dnet(gen, labels).mean()
            loss_g.backward(); opt_g.step()
            g_losses.append(loss_g.item())

    # ---------- held-out evaluation ----------
    eval_metrics = {}
    do_eval = (epoch % CONFIG["fid_frequency"] == 0) or (epoch == CONFIG["epochs"])
    if do_eval:
        G.eval()
        eval_metrics = evaluator.evaluate(sample_generator, n_gen=CONFIG["eval_n_gen"],
                                          batch=CONFIG["batch_size"], compute_diversity=True)

    # ---------- fixed conditional sample grid (balanced across classes) ----------
    G.eval()
    with torch.no_grad():
        grid = G(fixed_z, fixed_labels)
    viz.save_sample_grid(grid, os.path.join(paths.samples_dir, f"epoch_{epoch:04d}.png"),
                         title=f"cWGAN-GP epoch {epoch} (rows=classes)", nrow=CONFIG["num_classes"])

    epoch_time = time.time() - t0
    gpu_mb = env.gpu_memory_mb()
    row = {
        "epoch": epoch,
        "generator_loss": float(np.mean(g_losses)) if g_losses else float("nan"),
        "critic_loss": float(np.mean(c_losses)),
        "gradient_penalty": float(np.mean(gps)),
        "wasserstein_estimate": float(np.mean(wass)),
        "real_accuracy": float("nan"),   # N/A for a WGAN critic (not fabricated)
        "fake_accuracy": float("nan"),   # N/A for a WGAN critic (not fabricated)
        "fid": eval_metrics.get("fid", float("nan")),
        "kid_mean": eval_metrics.get("kid_mean", float("nan")),
        "kid_std": eval_metrics.get("kid_std", float("nan")),
        "diversity": eval_metrics.get("diversity", float("nan")),
        "lr_g": opt_g.param_groups[0]["lr"], "lr_d": opt_d.param_groups[0]["lr"],
        "epoch_time": epoch_time, "gpu_memory_mb": gpu_mb,
    }
    history.append(row)

    payload = {
        "epoch": epoch, "G": G.state_dict(), "D": Dnet.state_dict(),
        "opt_g": opt_g.state_dict(), "opt_d": opt_d.state_dict(),
        "rng": capture_rng_states(), "history": history.rows,
        "config": CONFIG, "best_metric": ckpt.best_metric,
    }
    saved = ckpt.save(epoch, payload, metric=row["fid"] if do_eval else None)

    T.epoch_summary_print(
        "cWGAN-GP", epoch, CONFIG["epochs"],
        {"Generator Loss": round(row["generator_loss"], 4),
         "Critic Loss": round(row["critic_loss"], 4),
         "Gradient Penalty": round(row["gradient_penalty"], 4),
         "Wasserstein Estimate": round(row["wasserstein_estimate"], 4),
         "Real Accuracy": "N/A (WGAN critic)", "Fake Accuracy": "N/A (WGAN critic)"},
        eval_metrics,
        {"Generator": row["lr_g"], "Critic": row["lr_d"]},
        epoch_time, saved, gpu_mb)

print("Training complete.")
```

## 18. Final evaluation + per-motif samples

```python

# =====================================================================
# FINAL EVALUATION (best checkpoint) + PER-MOTIF generated samples
# =====================================================================
best_state = ckpt.load(ckpt.best_path, map_location=DEVICE) or ckpt.load(ckpt.latest_path, map_location=DEVICE)
if best_state is not None:
    G.load_state_dict(best_state["G"]); best_epoch = int(best_state.get("epoch", -1))
else:
    best_epoch = -1
G.eval()

final = evaluator.evaluate(sample_generator, n_gen=CONFIG["eval_n_gen"],
                           batch=CONFIG["batch_size"], compute_diversity=True)

# per-motif sample grids (generate ALL classes)
for ci, motif in enumerate(D.MOTIF_ORDER):
    z = torch.randn(16, CONFIG["latent_dim"], device=DEVICE)
    lab = torch.full((16,), ci, dtype=torch.long, device=DEVICE)
    with torch.no_grad():
        imgs = G(z, lab)
    safe = motif.replace(" ", "_").replace("/", "-")
    viz.save_sample_grid(imgs, os.path.join(paths.samples_dir, f"final_class_{ci}_{safe}.png"),
                         title=f"cWGAN-GP — {motif}")

hist_df = pd.read_csv(paths.history_csv)
best_row = hist_df.loc[hist_df["fid"].idxmin()] if hist_df["fid"].notna().any() else None
final_metrics = {
    "model": "cWGAN_GP",
    "best_epoch": int(best_row["epoch"]) if best_row is not None else best_epoch,
    "best_fid": float(best_row["fid"]) if best_row is not None else float("nan"),
    "best_kid": float(best_row["kid_mean"]) if best_row is not None else float("nan"),
    "final_fid": final["fid"], "final_kid": final["kid_mean"], "final_kid_std": final["kid_std"],
    "final_diversity": final["diversity"],
    "final_generator_loss": float(hist_df["generator_loss"].iloc[-1]),
    "final_critic_loss": float(hist_df["critic_loss"].iloc[-1]),
    "final_discriminator_loss": float(hist_df["critic_loss"].iloc[-1]),  # alias for comparison
    "final_gradient_penalty": float(hist_df["gradient_penalty"].iloc[-1]),
    "training_time_sec": float(hist_df["epoch_time"].sum()),
    "peak_gpu_memory_mb": float(hist_df["gpu_memory_mb"].max()),
    "generator_parameters": CONFIG.get("generator_parameters"),
    "discriminator_parameters": CONFIG.get("discriminator_parameters"),
    "n_real_test": int(real_uint8.shape[0]), "n_gen_eval": CONFIG["eval_n_gen"],
    "real_accuracy": "N/A (WGAN critic has no accuracy)",
    "fake_accuracy": "N/A (WGAN critic has no accuracy)",
    "generator_accuracy": "N/A (not defined for GANs)",
}
with open(paths.final_metrics_json, "w") as f:
    json.dump(final_metrics, f, indent=2, default=str)
pd.DataFrame([final_metrics]).to_csv(paths.final_metrics_csv, index=False)
with torch.no_grad():
    viz.save_sample_grid(G(fixed_z, fixed_labels),
                         os.path.join(paths.samples_dir, "final_grid.png"),
                         title=f"cWGAN-GP final (best epoch {final_metrics['best_epoch']})",
                         nrow=CONFIG["num_classes"])
print(json.dumps(final_metrics, indent=2, default=str))
```
